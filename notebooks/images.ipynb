```python
img = cv2.imread('/home/mahad/projects/computer_vision_learning/data/resized_dog.jpeg')

# Select the region with your mouse
x, y, w, h = cv2.selectROI("Select the Dog", img)

# Crop the selected region
cropped = img[y:y+h, x:x+w]

cv2.imshow("Cropped Object", cropped)

cv2.waitKey(0)
cv2.destroyAllWindows()

# Select a ROI and then press SPACE or ENTER button!
# Cancel the selection process by pressing c button!
```

### Video


```python
video = cv2.VideoCapture("/home/mahad/projects/computer_vision_learning/data/FPV Drone Flight through Beautiful Iceland Canyon.mp4")

# Read and display frames
while video.isOpened():
    success, frame = video.read()

    if not success:
        break

    cv2.imshow("My First Video", frame)

    # Press q to stop the video
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

# Release resources
video.release()
cv2.destroyAllWindows()
```

### reading a Webcam

```python
webcam = cv2.VideoCapture(0)

while True:
    success, frame = webcam.read()
    if not success:
        break
    cv2.imshow("My Webcam", frame)
    if cv2.waitKey(40) & 0xFF == ord('q'):
        break
webcam.release()
cv2.destroyAllWindows()
```
